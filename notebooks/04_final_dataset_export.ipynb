# BloodBridge AI — Final Feature-Engineered Dataset Export

In this notebook, we structure the final feature engineered dataset by selecting only the columns needed for training our ML models, validating schemas, and writing the final dataset file.

```python
from pathlib import Path
import numpy as np
import pandas as pd

# Locate project root
ROOT = Path.cwd().resolve()
while ROOT != ROOT.parent and not (ROOT / "src").exists():
    ROOT = ROOT.parent
    
DATA_DIR = ROOT / "data"
print(f"Project root identified: {ROOT}")
```

### Load the Feature Engineered Dataset

```python
input_path = DATA_DIR / "processed" / "feature_engineering_working_synthetic.csv"
featured_df = pd.read_csv(input_path, parse_dates=["record_date"])
print(f"Loaded dataset with {len(featured_df):,} rows.")
```

### Column Selection and Final Schema Structuring

We select the target prediction column `shortage_next_day`, time indicators, location information, and our rolling window lag features.

```python
target_columns = [
    "record_date",
    "bank_id",
    "city",
    "blood_group",
    "inventory_ratio",
    "day_of_week",
    "month",
    "emergency_event_flag",
    "requests_received_rolling_7d",
    "donations_received_rolling_7d",
    "inventory_ratio_rolling_7d",
    "request_donation_gap_7d",
    "shortage_next_day",
    "data_origin",
    "generation_seed"
]

final_df = featured_df[target_columns].copy()
print(f"Final output dataset columns: {len(final_df.columns)}")
print("\nFirst 3 rows of restructured dataset:")
display(final_df.head(3))
```

### Export Final Feature-Engineered Dataset

```python
output_path = DATA_DIR / "processed" / "final_feature_engineered_bloodbridge_dataset.csv"
final_df.to_csv(output_path, index=False)
print(f"Successfully exported {len(final_df):,} rows to {output_path.resolve()}")
```
